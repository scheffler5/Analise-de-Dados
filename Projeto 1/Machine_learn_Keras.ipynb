```python
import warnings
warnings.filterwarnings("ignore")
import pandas as pd
```

```python
data = pd.read_csv('atlas2010_recorte.csv', delimiter=',')
```

```python
X = data[['mort1', 'razdep', 'e_anosestudo']]
y = data[['fectot']]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)
```

```python
from keras.models import Sequential
from keras.layers import Dense
from keras.optimizers import Adam
```

### Rodando o primeiro modelo de rede neural

```python
model = Sequential()
model.add(Dense(10, input_dim=3, activation='linear'))
model.add(Dense(6, activation='relu'))
model.add(Dense(1, activation='linear'))
```

```python
model.compile(loss='mse', optimizer=Adam(learning_rate=0.1), metrics=['mse'])
```

```python
model.fit(X_train, y_train, epochs=30, batch_size=40, verbose=1)
```

```python
model.summary()
```

```python
from sklearn.metrics import r2_score

# Obter as previsões do modelo para os dados de teste
y_pred = model.predict(X_test)

# Calcular o coeficiente de determinação R²
r2 = r2_score(y_test, y_pred)

print(r2)
```
