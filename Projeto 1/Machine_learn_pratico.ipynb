```python
import numpy as np 

#definição da arquitetura da rede neural
input_size = 2
hidden_size = 4
output_size = 1

#carregamento dos dados de treinamento e teste
x_train = np.array([[0,0],[0,1],[1,0],[1,1]])
y_train = np.array([[0],[1],[1],[0]])
x_test = np.array([[0,0],[0,1],[1,0],[1,1]])
y_test = np.array([[0],[1],[1],[0]])
```

```python
# atribuição de pesos aleatorios
w1 = np.random.randn(input_size,hidden_size)
b1 = np.zeros((1,hidden_size))
w2 = np.random.randn(hidden_size, output_size)
b2 = np.zeros((1, output_size))

#Definição da função de ativação sigmoid
def sigmoid (x):
    return 1 / (1 + np.exp(-x))

def mse_loss(y_true, y_pred): 
    return ((y_true - y_pred) **2).mean()

def sgd(params, gradients, learning_rate):
    for param, gradient in zip(params, gradients):
        param -= learning_rate * gradient
        
```

```python
# execucao do treinamento da rede neural
num_epochs = 10000
learning_rate = 0.1

for epoch in range(num_epochs):
    z1 = np.dot( x_train, w1) + b1
    a1 = sigmoid(z1)
    z2 = np.dot(a1,w2) + b2
    y_pred = sigmoid(z2)

    loss = mse_loss(y_train, y_pred)
    accuracy = ((y_pred > 0.5) == y_train).mean()


    delta2 = (y_pred - y_train) * (y_pred* (1 - y_pred))
    delta1 = delta2.dot(w2.T) * (a1 * (1 - a1))
    dw2 = np.dot(a1.T, delta2)
    db2 = delta2.sum(axis=0, keepdims= True)
    dw1 = np.dot(x_train.T, delta1)
    db1 = delta1.sum(axis=0)

    sgd([w1,b1,w2,b2], [dw1,db1,dw2,db2], learning_rate)

    if epoch % 1000 == 0:
        print (f"Epoch {epoch}: loss = {loss:.4f}, accuracy = {accuracy:.4f}")
        
```
