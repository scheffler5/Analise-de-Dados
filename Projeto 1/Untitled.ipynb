Em seguida, aplicará uma equação que determine, com algum grau de aleatoriedade, os pesos dos indivíduos. Logo depois, calculará o IMC e definirá para cada indivíduo a existência (sim ou não) de problemas cardiácos, ambos com algum grau mnimo de correspondência com o IMC.

Ao final, faça as análises necessárias com medidas de dispersão e de tendência central. Crie as funções para tal. E interprete os IMCs obtido conforme WHO (Organização Mundial da Saúde).

Os dados devem estar reunidos em uma tabela. Todas as técnicas aplicadas aqui são vistas na Aula 22.

#### ***Definindo o tamanho da amostra***

```python
Tamanho_amostra = 10000
```

#### ***Realizando as Importaçoes dos Pacotes***


```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from sympy import *
from random import uniform
```

#### ***Gerando Alturas aleatorias***

```python
altura = np.random.normal(1.75, 0.3, Tamanho_amostra)
altura = altura.tolist()
```

#### ***Gerando o peso a partir das alturas***

```python
peso=[]
for i in altura:
    a = round(float(uniform(-5, 5)), 2)
    b = 50*i
    c = a + b
    peso.append(c)
len(peso)
```

#### ***Gerando o IMC***

```python
IMC=[]
for i,j in zip(peso,altura):
    e = i / (j**2)
    e = round(e, 2)
    IMC.append(e)
```

### ***Gerando os sexos de forma aleatória***

```python
SEXO = []
for i in range (0,10000):
    value = np.random.uniform(0,1)
    if value > 0.5:
        SEXO.append('M')
    else:
        SEXO.append('F')
```

### ***Gerando a correspondência com problemas cardíacos***

```python
CARDIACO = []
for i in range (0,10000):
    if IMC[i] > 40:
        CARDIACO.append('S')
    else:
        if IMC[i] > 30 and IMC[i] < 40:
            CARDIACO.append('P')
        else:
            CARDIACO.append('N')
```

#### ***Gerando o DataFrame***

```python
z = altura,peso,IMC,SEXO,CARDIACO
```

```python
df = pd.DataFrame(z, index=['Altura', 'Peso', 'IMC', 'SEXO','CARDIACO']).T
```
